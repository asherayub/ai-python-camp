# Calculator using python

```python
num1 = int(input('Enter first number: '))
num2 = int(input('Enter second number: '))

operator = input('Enter operator (+, -, /, *, ^): ')

if operator == '+':
    print(num1,operator,num2, '=', num1 + num2)
elif operator == '-':
    print(num1,operator,num2, '=', num1 - num2)
elif operator == '/':
    print(num1,operator,num2, '=', num1 / num2)
elif operator == '^':
    print(num1,operator,num2, '=', num1 ** num2)
```

# Check if there is a numeric value in a list using for loop

```python
list = ['abc', 'exf', 'soxk', 3, 'skei']

for val in list:
    if type(val) is int:
        print('List contains one or more integer value(s)')
```

# Sum all items in numeric dictionary

```python
fruit_prices = {
    'apple': 200,
    'orange': 420,
    'banana': 230,
    'kiwi': 500,
    'dragon fruit': 1200
}

print('Total is:', sum(fruit_prices.values()))
```

# Identify duplicate values in a list

```python
my_list = ['car', 'bike', 'jet', 'bicycle', 'chopper', 'jet', 'bike', 'truck', 'car', 'bike']

duplicates = []
checked = []
for item in my_list:
    if item in checked:
        if item in duplicates:
            continue
        duplicates.append(item)
    else:
        checked.append(item)

print(duplicates)
```

# Check if a given key already exists in a dictionary

```python
fruit_prices = {
    'apple': 200,
    'orange': 420,
    'banana': 230,
    'kiwi': 500,
    'dragon fruit': 1200,
}

key_to_check = input('What key are you looking to find a duplicate for?: ')

if key_to_check in fruit_prices:
    print('Key exists in the current dictionary')
else:
    print('Key exists in the current dictionary')
```
